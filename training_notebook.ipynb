```python
from sklearn.datasets import load_diabetes   #returns bunch not pandas dataframe
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
diabetes=load_diabetes()
```

```python
print(diabetes["DESCR"])
```

```python
df = pd.DataFrame(
    diabetes.data,
    columns=diabetes.feature_names
)
```

```python
df.head()
```

```python
X=df
y=diabetes["target"]
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.30,random_state=42)
```

```python
X_train.corr()
```

```python
plt.figure(figsize=(15,10))
sns.heatmap(X_train.corr(),annot=True)
```

```python
from sklearn.tree import DecisionTreeRegressor
model=DecisionTreeRegressor()
```

```python
params={
    "criterion" : ["squared_error", "absolute_error", "poisson"],
    "splitter" : ["best", "random"],
    "max_depth" : [x for x in range(1,100)]  , #x=0 is not valid is max depth
    "max_features" : ["sqrt", "log2" , "auto"]
}
```

```python
from sklearn.model_selection import GridSearchCV
grid=GridSearchCV(model,param_grid=params , cv=5 , scoring="neg_mean_squared_error")

import warnings
warnings.filterwarnings("ignore")

grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
y_pred=grid.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error

```

```python
print(r2_score(y_test,y_pred))
print(mean_absolute_error(y_test,y_pred))
print(mean_squared_error(y_test,y_pred))
      
```

# SELECTED MODEL

```python
selected=DecisionTreeRegressor(criterion="poisson",max_depth=5,max_features="log2" , splitter="best")
selected.fit(X_train,y_train)
```

```python
from sklearn import tree
```

```python
plt.figure(figsize=(15,10))
tree.plot_tree(selected,filled=True)
```

```python
plt.figure(figsize=(30, 20))
tree.plot_tree(
    selected,
    filled=True,
    rounded=True,
    fontsize=10
)

plt.savefig("decision_tree.png", dpi=300, bbox_inches="tight")
plt.show()
```
